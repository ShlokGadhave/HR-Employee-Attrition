```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns
```

Basic Understanding 

```python
df=pd.read_csv('HR-Employee-Attrition.csv')
print(df.head())
```

```python
#Data Cleaning 
df.dropna(inplace=True)
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
#dropping columns 
df.drop(columns=['EmployeeCount',
'EmployeeNumber',
'Over18',
'StandardHours'],inplace=True)
```

```python
#Type-Casting Method 
columns=['Attrition',
            'BusinessTravel',
            'Department',
            'EducationField',
            'Gender',
            'JobRole',
            'MaritalStatus',
            'OverTime']
for cols in columns:
    df[cols]=df[cols].astype('category')
df.dtypes
```

```python
rating_cols = ['Education',
               'EnvironmentSatisfaction',
               'JobInvolvement',
               'JobLevel',
               'JobSatisfaction',
               'PerformanceRating',
               'RelationshipSatisfaction',
               'StockOptionLevel',
               'WorkLifeBalance']

for col in rating_cols:
    df[col] = df[col].astype('category')
df.dtypes
```

```python
#Feature Engineering 
#age grp function 
df['Age Group']=pd.cut(df['Age'],bins=[18,25,35,45,60],labels=['18-25','26-35','36-45','46-60'])
df['Age Group']
```

```python
df['MonthlyIncome'].head(10)
df['IncomeRate']=pd.cut(df['MonthlyIncome'],bins=[1000,5000,10000,15000,20000,25000],labels=['VeryLow','Low','Medium','Average','High'])
df['IncomeRate'].head()
```

```python
df['TotalWorkingYears'].head(10)
df['Experience']=pd.cut(df['TotalWorkingYears'],bins=[0,5,10,15,20,25],labels=['Fresher','MidLevel','Expert','SeniorEmployee','Manager'])
df['Experience'].head(10)
```

```python
df['PromotionGap'] = df['YearsAtCompany'] - df['YearsSinceLastPromotion']
df['PromotionGap'].head(10)
```

```python
df['Distancefromoffice']=pd.cut(df['DistanceFromHome'],bins=[0,5,10,15,20,25],labels=['Nearby','Mid','CanTravel','Far','TooFar'])
df['Distancefromoffice'].head(10)
```

```python
df['JobLevel']=df['JobLevel'].astype('int')
df['IncomePerLevel'] = df['MonthlyIncome'] / df['JobLevel']
df['IncomePerLevel'].head(10)

```

```python
#OverTime Flag 
df['OverTimeFlag']=df['OverTime'].map({'Yes':1,'No':0})
df['OverTimeFlag']
```

```python
df['AttritionFlag'] = df['Attrition'].map({'Yes':1,'No':0})
df['AttritionFlag']
```

```python
df['BusinessTravel'] = df['BusinessTravel'].map({
    'Non-Travel':1,
    'Travel_Rarely':2,
    'Travel_Frequently':3
})

df['DistanceFromHome'] = df['DistanceFromHome'].astype(int)

df['BusinessTravel'] = df['BusinessTravel'].map({
    'Non-Travel':1,
    'Travel_Rarely':2,
    'Travel_Frequently':3
})

df['TravelStressScore'] = (
    df['DistanceFromHome'] *
    df['BusinessTravel']
)
df['TravelStressScore'].head(10)
```

```python
#Visualization
plt.figure(figsize=(6,4))
sns.histplot(x='Age',data=df,hue='Attrition',palette="Set2")
plt.title('Age Disturbution')
plt.show()
```
